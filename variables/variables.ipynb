```python
## Understanding Variable Types
## Python is dynamically typed, type of a variable is determined at runtime
age=25 #int
height=6.1 #float
name="Lamech" #str
is_student=True #bool

print(type(name))
print(type(height))
print(type(age))
```

```python
# Type Conversion
age=25
print(type(age))

age_str=str(age)
print(age_str)
print(type(age_str))
```

```python
## Dynamic Typing
## Pythion allows the type of a variable to change during runtime
var = 10 #int
print(var, type(var))
var = "Hello, World!" #str
print(var, type(var))

var=3.14
print(var, type(var))
```

```python
## Simple Calculator
num1=int(input("Enter first number: "))
num2=int(input("Enter second number: "))

sum = num1 + num2
difference = num1 - num2
product = num1 * num2
quotient = num1 / num2

print("Sum:", sum)
print("Difference:", difference)
print("Product:", product)
print("Quotient:", quotient)
```
